```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# texto = "Estar em toda a parte onde houver bem a fazer e sofrimento a aliviar"

texto = 'PDF significa Portable Document Format (Format de Documento Portátil).' \
'É um formato de arquivo criado pela Adobe em 1993 para partilhar documentos sem perder a formatação original'

# Conjunto de palavras que vão ser removidas do texto

as_minhas_stopwords = set(['em' , 'toda' , 'a' , 'onde' , 'e'])

# ---------------------------- Configurar e gerar a nuvem a partir do texto -----------------------

nuvem = WordCloud (
    width=1000, 
    height=600,
    background_color="#F8F3F8",
    colormap='Greens', # outras cores disponiveis na clasroom

# 'plasma' , 'Blues' , 'Greens'

    max_words=20, # nº maximo de palavras no cartaz
    stopwords=as_minhas_stopwords # remover as minhas palavras
).generate(texto)

# ----------------------- Exibir a imagem através da Matplotlib (biblioteca python usada para criar gráficos) ---------------

plt.figure(figsize=(10, 5)) # Tamanho da exibição da figura
plt.imshow(nuvem, interpolation = 'bilinear') # Renderiza a imagem
plt.axis('off')
plt.show

nuvem.to_file('nuvem_palavras.png')
```
